```python
import yfinance as yf
import matplotlib.pylab as plt
```

```python
bitcoin =yf.Ticker("BTC-USD")
```

```python
bit_data = bitcoin.history(period='max')
bit_data
```

```python
bit_data.tail()
```

```python
point = bit_data[(bit_data["Close"] > 30000) & (bit_data["Close"] < 60000)]
point
```

```python
plt.figure(figsize=(12, 6))
plt.plot(point.index,point["Close"],color="Orange",linewidth=1)
plt.title("Bitcoin Price between $30,000 & $60,000", fontsize=12)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Close Price (USD)", fontsize=12)
plt.grid(True)
plt.tight_layout()
```

```python
plt.figure(figsize=(14,6))
plt.plot(bit_data.index,bit_data["Close"],color="Orange",linewidth=1.5)
ath_value = bit_data["Close"].max()
ath_date = bit_data["Close"].idxmax()

plt.scatter(ath_date, ath_value, color='red', zorder=6)
plt.text(ath_date, ath_value, f'ATH: ${ath_value:,.0f}\n({ath_date.date()})', 
         color='red', fontsize=9, ha='right', va='top')
plt.title("Bitcoin History")
plt.xlabel("Date",fontsize=10)
plt.ylabel("Bitcoin Close Price",fontsize=10)
plt.grid(True)
plt.tight_layout()
```

```python
print("Bitcoin history statistics")
print(f'Bitcoin All Time Max: {bit_data["Close"].max():,.2f} USD')
print(f'Bitcoin All Time Low: {bit_data["Close"].min():,.2f} USD')
print(f'Bitcoin Mode (close): {bit_data["Close"].mode()[0]:,.2f} USD')
```

```python
import seaborn as sns
Corr = bit_data[["Open","High","Low","Close","Volume"]].corr()
Corr
```

```python
"""These coefficients range from -1 to 1:

1: Perfect positive correlation

0: No correlation

-1: Perfect negative correlation"""


plt.figure(figsize=(8, 6))   
sns.heatmap(Corr,annot=True,fmt=".2f",cmap="coolwarm",linewidths=1.5)           
plt.title("Bitcoin Feature Correlation Heatmap")
plt.tight_layout()
plt.show()
```
